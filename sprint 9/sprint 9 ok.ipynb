Función para definir las franjas horarias(asignar_franja) usando una condicion if,definiendo cuatros franjas horarias Madrugada,Mañana,Trade y Noche. 
Aplicacamos con la funcion apply la funcion  asignar_franja y creamos un campo nuevo en df_transactions llamado 'franja_horaria'.

```python
def asignar_franja(fecha):
    hora = fecha.hour
    if 0 <= hora < 6:
        return '01. Madrugada (00:00 - 05:59)'
    elif 6 <= hora < 12:
        return '02. Mañana (06:00 - 11:59)'
    elif 12 <= hora < 18:
        return '03. Tarde (12:00 - 17:59)'
    else:
        return '04. Noche (18:00 - 23:59)'

df_transactions['franja_horaria'] = df_transactions['timestamp'].apply(asignar_franja)
```

Construimos la tabla agregada utilizando crosstab (se cruzan las dos variables Filas: Franjas, Columnas: Años) y se suman las transccciones por franja horaria en ese año

```python
tabla_pivot_horas = pd.crosstab(
    index=df_transactions['franja_horaria'],
    columns=df_transactions['year'],
    values=df_transactions['amount'],
    aggfunc='sum'
).fillna(0)

print("--- Tabla de Facturación por Franja Horaria y Año ---")
print(tabla_pivot_horas)
```

 --- CONTINUACIÓN DEL ANÁLISIS ESTÁNDAR ---
 1 Combinar con la tabla de maestros de productos
 2 Identificar la primera compra absoluta de cada usuario en el sistema
 3 Calcular variabilidad utilizando el monto prorrateado por producto
 4 Agregar métricas por producto
 5 Clasificación mediante medianas

```python
# 1
df_combinado = pd.merge(df_transactions_exploded, df_products,left_on='product_ids',right_on='id', how='inner')
df_combinado['timestamp'] = pd.to_datetime(df_combinado['timestamp'])

# 2
df_primera_compra = df_combinado.groupby('user_id')['timestamp'].min().reset_index()
df_primera_compra = df_primera_compra.rename(columns={'timestamp': 'fecha_primera_compra'})

df_combinado = pd.merge(df_combinado, df_primera_compra, on='user_id', how='left')
df_combinado['es_nou_usuari'] = df_combinado['timestamp'] == df_combinado['fecha_primera_compra']

# 3
df_var = df_combinado.groupby('product_name')['amount_prorrateado'].agg(
    lambda x: x.std() / x.mean() if x.mean() != 0 else 0
).reset_index(name='variabilidad_ingresos')

# 4
df_prod_metrics = df_combinado.groupby('product_name').agg(
    ingresos_totales=('amount_prorrateado', 'sum'), # Usamos el ingreso corregido
    nuevos_usuarios=('es_nou_usuari', 'sum'),
    total_compradores_unicos=('user_id', 'nunique')
).reset_index()

df_productos = pd.merge(df_prod_metrics, df_var, on='product_name')
df_productos['porcentaje_nuevos'] = (df_productos['nuevos_usuarios'] / df_productos['total_compradores_unicos']) * 100

# 5
mediana_ingresos = df_productos['ingresos_totales'].median()
mediana_captacion = df_productos['nuevos_usuarios'].median()

def clasificar_producto(row):
    if row['ingresos_totales'] >= mediana_ingresos and row['nuevos_usuarios'] >= mediana_captacion:
        return 'Captador'
    elif row['ingresos_totales'] >= mediana_ingresos and row['nuevos_usuarios'] < mediana_captacion:
        return 'Recurrente'
    elif row['ingresos_totales'] < mediana_ingresos and row['nuevos_usuarios'] >= mediana_captacion:
        return 'Emergente'
    else:
        return 'Estancado'

df_productos['tipologia'] = df_productos.apply(clasificar_producto, axis=1)
```
