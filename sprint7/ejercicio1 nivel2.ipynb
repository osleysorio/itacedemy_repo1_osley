## Importar librerias

```python
import pandas as pd
import numpy as np
```

## Lista real de pacientes

```python
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
```

 ## Diccionario con niveles de riesgos

```python
nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

## Funcion para convertir cada string en una estructura limpia (diccionario o similar).

```python
def separar_datos(lista_pacientes):
    pacientes_estructurados = []
    for paciente in lista_pacientes:
        # Dividir los tres bloques principales por la barra '|'
        bloques = paciente.split('|')

        # Guardar en un diccionario base manteniendo los textos originales
        pacientes_estructurados.append({
            "nombre": bloques[0],
            "edad": int(bloques[1]),
            "sintomas": bloques[2]
        })
    return pacientes_estructurados
```

## Funcion para separar y normalizar la la lista de sintomas.
Separar y normalizar la lista de síntomas:
minúsculas
eliminar espacios
dividir correctamente por comas

```python
# FUNCIÓN 2: Limpiar, estandarizar y normalizar los síntomas con el diccionario de riesgo
def limpiar_lista(lista_estructurada):
    lista_limpia = []
    for p in lista_estructurada:
        # Estandarizar nombre (minúsculas y sin espacios)
        nombre_limpio = p["nombre"].strip().lower()

        # Procesar la cadena de síntomas: separar por comas, limpiar espacios y pasarlo a minúscula
        sintomas_lista = [s.strip().lower().replace(" ", "_") for s in p["sintomas"].split(',')]

        # Construir el diccionario final limpio
        lista_limpia.append({
            "nombre": nombre_limpio,
            "edad": p["edad"],
            "sintomas": sintomas_lista,

        })
    return lista_limpia
```

 ## Funcion para asignar un riesgo a cada sintoma

 Asignar un riesgo a cada síntoma utilizando el diccionario nivells_risc.



```python
def normalizar(lista_estructurada, diccionario_riesgo):
    for p in lista_estructurada:
        # 1. Extraer la lista de síntomas que tiene este elemento actual
        sintomas_lista = p.get("sintomas", [])

        # 2. Mapear a valores numéricos usando el diccionario de riesgo
        sintomas_numericos = []
        for clave_sintoma in sintomas_lista:
            valor = diccionario_riesgo.get(clave_sintoma, 0)
            sintomas_numericos.append(valor)

        # 3. AGREGAR las nuevas claves al diccionario del elemento actual
        p["valores_riesgo"] = sintomas_numericos
        p["riesgo_total"] = sum(sintomas_numericos)

    # Retorna la lista con los diccionarios originales ya modificados
    return lista_estructurada
```

##Funcion
Ordenar a los pacientes de mayor a menor riesgo para determinar la prioridad de atención.

```python
def ordenar_por_riesgo(lista_estructurada):
    return sorted(lista_estructurada, key=lambda x: x["riesgo_total"], reverse=True)
```

```python
def main():
  datos_separados = separar_datos(pacients)
  datos_limpios = limpiar_lista(datos_separados)
  datos_con_riesgo = normalizar(datos_limpios, nivells_risc)
  pacientes_ordenados = ordenar_por_riesgo(datos_con_riesgo)
  return pacientes_ordenados

pacientes_ordenados = main()
print(pacientes_ordenados)
```
