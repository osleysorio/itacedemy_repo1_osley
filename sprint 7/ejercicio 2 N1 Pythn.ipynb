## Convertir  datos del cliente

- Convierte a un dataframe la lista de clientes que esta en una lista de de dicionarios

```python
def convertir_dataframe(lista_de_diccionarios):
  df = pd.DataFrame(lista_de_diccionarios)
  return df
```

# Funcion asignar factor de codigo
- creamos un dicionario que contiene las equivalencias entre el comportamiento y el factor de codigo
- Asiganamos una nueva columna al dataframe con esos factores de codigo por cliente  a traves de un mapeo, asigamos valor 0 por defecto para evitar si hay algun valor fuera de lso parametros.

```python
def asignar_factor_codigo(df):

    # 1. Definimos el diccionario con las equivalencias
    mapeo = {
        0: 0.1,
        1: 0.25,
        2: 0.5,
        3: 0.75,
        4: 0.9
    }

    # 2. Aplicamos el mapeo y rellenamos los valores no encontrados con el valor por defecto
    df['factor_codigo'] = df['codi_comportament'].map(mapeo).fillna(0.0)

    return df
```

#Funcion obtenir_factor_edat

- creamos un dicionario que contiene las equivalencias entre el los grupss de edad y  factor de de edad
- Asiganamos una nueva columna al dataframe con esos factores de edad por cliente a traves de un mapeo, asigamos valor 0 por defecto para evitar si hay algun valor fuera de los parametros.

```python
def asignar_factor_edad(df):

    # 1. Definimos el diccionario con las equivalencias
    mapeo_edad = {
        '18-25': 1.25,
        '26-40': 1,
        '41-60': 0.85,
        '60+': 0.7,

    }

    # 2. Aplicamos el mapeo y rellenamos los valores no encontrados con el valor por defecto
    df['factor_edad'] = df['grup_edat'].map(mapeo_edad).fillna(0.0)

    return df
```

#Funcion calcular riesgo final

-calcula el riesgo multiplicando el el factor de codigo * el factor de edad

```python
def calcular_riesgo_final(df):

    df['risc_final'] = df['factor_codigo'] * df['factor_edad']

    return df

```

## Funcion para segmentar los clientes segun el riesgo final
- (valores altos de risc_final significa cliente de alto potencial o bajo riesgo)
- (valores bajos de risc_final significa cliente de bajo potencial o alto riesgo)
- definimos condiciones basados en la la colunna risc_final y tomando valores de referencia para compara con los valores dela columna
- definimos las etiquetas que vamos a utilizar para segmentar  los cientes segun su potencial
- aplicamos np.select con estas confcione spara que claisfique segun las etiquetas

```python

def clasificar_segmento_cliente(df):

    # 1. Definimos las condiciones basadas en la columna del DataFrame recibido
    condiciones = [
        df['risc_final'] > 0.75,
        (df['risc_final'] <= 0.75) & (df['risc_final'] > 0.50),
        df['risc_final'] <= 0.50
    ]

    # 2. Definimos las etiquetas correspondientes
    clases = ['Alto', 'Medio', 'Bajo']

    # 3. Aplicamos np.select de forma vectorizada
    df['seg_cliente_potencial'] = np.select(condiciones, clases, default='Bajo')

    return df
```

# Funcion processar_clients
-obtiene los datos de las funciones:
    - obtenir_segment_i_factor
    - obtenir_factor_edat
    - obtenir_calcular_risc
    

```python
# Funcion processar_clients (Orquestadora de transformación de datos)
def processar_clients(lista_de_diccionarios):
    df = convertir_dataframe(lista_de_diccionarios)
    df = asignar_factor_codigo(df)
    df = asignar_factor_edad(df)
    df = calcular_riesgo_final(df)
    df = clasificar_segmento_cliente(df)
    return df
```

## funcion mostrar_dataframe
- muestra infromacion en un dataframe

```python
## Funcion mostrar_dataframe
def mostrar_dataframe(df):
    print("--- DATAFRAME COMPLETO ---")
    print(df.to_string(index=False))
    print("\n" + "="*50 + "\n")
```

## Funcion mostrar_respostes

- Muestra respuesta
1. ¿Qué clientes deben contactar primero en una campaña de retargeting?
2. ¿Qué diferencias existen entre clientes con el mismo comportamiento pero de edades distintas? (Por ejemplo: ¿un cliente joven que abandona la cesta de compra hace 24 h tiene el mismo riesgo que uno de más edad?)
3. ¿Qué clientes presentan tan poco interés que no vale la pena invertir presupuesto?
4. ¿Qué cliente sería lo más rentable de recuperar?

```python
def mostrar_info(df):
  # 1 muestra clientes que primero se deben contactar en una campaña
  print(df[df['seg_cliente_potencial'] == 'Alto'][['id_client', 'nom', 'risc_final']])
  print("\n" + "-"*50 + "\n")

  # 2 Por ejemplo: ¿un cliente joven que abandona la cesta de compra hace 24 h tiene el mismo riesgo que uno de más edad
  ejemplo_edad = df[df['codi_comportament'] == 4][['nom', 'grup_edat', 'factor_edad', 'risc_final', 'seg_cliente_potencial']].sort_values(by='risc_final', ascending=False)
  print(ejemplo_edad.to_string(index=False))
  print("Nota: A menor edad, el factor_edad es más alto (1.25). Por ello, un cliente joven que abandona la cesta tiene un 'risc_final' e interés potencial más alto que uno mayor.")
  print("\n" + "-"*50 + "\n")

  # 3 muestra clientes que no vale la pena invertir presupuesto
  print(df[df['seg_cliente_potencial'] == 'Bajo'][['id_client', 'nom', 'risc_final']])
  print("\n" + "-"*50 + "\n")

 # 4 muestra el cliente mas rentable a recuperar
  cliente_top = df.loc[[df['risc_final'].idxmax()]][['id_client', 'nom', 'risc_final']]
  print(cliente_top.to_string(index=False))
  print("\n" + "-"*50 + "\n")
```
