# Funcion para validar  conocimiento de catalan

1 Defino la funcion 'validar_catalan' que recibe un parametro(valor de conocimiento de catalan)

2 realizo un 'match case' para clasificar el valor que ha entrado como parametro

    case 0:  Sin conocimiento, entonces  retorno 0
    case 1:  hasta nivel B1 , entonces retorno 1
    case 2:  nivel B2, entonces retorno 2  
    case 3:  C1 o superior, entonces retorno 3
    case _:  Opción no válida, entonces retorno -1

  y retornamos un valor en en el caso que se cumpla.

```python

def validar_catalan(c_catalan):

  match int(c_catalan):

    case 0:
      print("Catalán: Sin conocimiento"); return 0
    case 1:
      print("Catalán: hasta nivel B1"); return 1
    case 2:
      print("Catalán: nivel B2"); return 2
    case 3:
      print("Catalán: C1 o superior"); return 3
    case _:
      print("Opción no válida"); return -1
```

# Funcion para validar  dominio de excel

1 defino la funcion 'validar_excel' que recibe un parametro(valor de conocimiento de excel)

2 realizo un 'match case' para clasificar el valor que ha entrado como parametro


    case 0:  Dominio excel:nivel básico, entonces  retorno 0
    case 1:  Dominio excel:uso de funciones , entonces retorno 1
    case 2:  Dominio excel:tablas dinámicas, entonces retorno 2  
    case 3:  Dominio excel:Visual Basic y Macros, entonces retorno 3
    case _:  Opción no válida, entonces retorno -1

  y retornamos un valor en en el caso que se cumpla.


```python
def validar_excel(d_excel):

  match int(d_excel):

    case 0:
      print("Dominio excel:nivel básico"); return 0
    case 1:
      print("Dominio excel:uso de funciones"); return 1
    case 2:
      print("Dominio excel:tablas dinámicas"); return 2
    case 3:
      print("Dominio excel:Visual Basic y Macro"); return 3
    case _:
      print("Opción no válida"); return -1
```

```python
def validar_dades(c_catalan, d_excel, t_academico):
    """
    Comprueba si las notas de competencias están dentro del rango exigido (0-3).
    Devuelve True si son válidas o False si hay algún error.
    """
    if not (0 <= c_catalan <= 3):
        return False
    if not (0 <= d_excel <= 3):
        return False
    if not (0 <= t_academico <= 3):
        return False
    return True
```

# Funcion para validar  titulo acedemico

1 defino la funcion 'validar_t_acedemico' que recibe un parametro(valor segun el titulo del candidato)

2 realizo un 'match case' para clasificar el valor que ha entrado como parametro


    case 0:  Titulo acedemico:ESO, entonces  retorno 0
    case 1:  Titulo acedemico: FP, entonces retorno 1
    case 2:  Titulo acedemico: Universidad, entonces retorno 2  
    case 3:  Titulo acedemico: Máster o superior, entonces retorno 3
    case _:  Opción no válida, entonces retorno -1

  y retornamos un valor en en el caso que se cumpla.

```python
def validar_t_acedemico(t_academico):

  match int(t_academico):

    case 0:
      print("Titulo acedemico:ESO"); return 0
    case 1:
      print("Titulo acedemico: FP"); return 1
    case 2:
      print("Titulo acedemico: Universidad"); return 2
    case 3:
      print("Titulo acedemico: Máster o superior"); return 3
    case _:
      print("Opción no válida"); return -1
```

# Funcion para calcular la cantidad de puntos obtenidos por candidato

1 defino funcion cal_puntuacion que recibe como parametros(expreincia, conocimiento de catala, domiminio de excel, titulo acedemico)
2 luego calculo la puntuacion usando la formula
  puntuacion = expreincia*1.5 + c_catalan + d_excel + t_acdemico
3 retorno el valor obtenido.

```python
def calcular_puntuacion (experiencia,c_catalan,d_excel,t_acedemico):
  return (experiencia * 1.5 + c_catalan + d_excel + t_acedemico)

```

#defino funcion que clasifique el candidato segun su puntiacion

if puntuación < 8 → No prioritario  
   sino is puntuación < 12  → Candidato adecuado
   sino Prioridad Alta

   debe devolver un nivel de prioridad por candidato

```python
def clasificar_candidato (puntuacion):

  if puntuacion < 8 :
    print ('No Prioritario')
    return 3
  elif puntuacion < 12 :
    print ('Aandidato Adecuado')
    return 2
  else:
    print ('Prioridad Alta')
    return 1
```

# Funcion obtener datos

```python
# Funcion obtener datos
def obtener_datos(nombre_apellidos,telefono,experiencia,c_catalan,d_excel,t_acedemico):

  #validar y obtenr puntos
  c_catalan = validar_catalan(c_catalan)
  d_excel = validar_excel(d_excel)
  t_acedemico = validar_t_acedemico(t_acedemico)
  # retornar lista con datos candidato
  return [nombre_apellidos,telefono,experiencia,c_catalan,d_excel,t_acedemico]
```

# Funcion crear informe

```python
def generar_informe(lista_candidatos):
    # Crear DataFrame con los campos solicitados
    df = pd.DataFrame(lista_candidatos, columns=['nombre_apellidos', 'teléfono', 'puntuación', 'prioridad'])

    # Ordenar primero por prioridad (Alta a No prioritario) y en caso de empate por puntuación (Mayor a menor)
    df_ordenado = df.sort_values(by=['prioridad', 'puntuación'], ascending=[True, False])

    # Eliminar la columna prioridad del reporte final si solo se piden los 3 campos del enunciado
    df_final = df_ordenado[['nombre_apellidos', 'teléfono', 'puntuación']]

    print("\n================ INFORME FINAL DE CANDIDATOS ================")
    print(df_final.to_string(index=False))
```

# Funcion principal

```python
def main():
    candidatos_inscritos = []
    MAX_CANDIDATOS = 10

    print(f"Inicio de captación. Registre a los {MAX_CANDIDATOS} candidatos.")

    for i in range(1, MAX_CANDIDATOS + 1):
        print(f"\n--- CANDIDATO {i} de {MAX_CANDIDATOS} ---")
        nombre = input("Nombre y Apellidos: ")
        telefono = input("Teléfono: ")
        experiencia = int(input("Años de experiencia laboral: "))
        c_catalan = int(input("Nivel de Catalán (0=Ninguno, 1=B1, 2=B2, 3=C1+): "))
        d_excel = int(input("Dominio de Excel (0=Básico, 1=Funciones, 2=Tablas, 3=VBA): "))
        t_academico =int( input("Título Académico (0=ESO, 1=FP, 2=Uni, 3=Master+): "))
        if c_catalan < 0 or c_catalan > 3 or d_excel < 0 or d_excel > 3 or t_academico < 0 or t_academico > 3:
          i = i
        else:
          # Procesar datos
          datos = obtener_datos(nombre, telefono, experiencia, c_catalan, d_excel, t_academico)

          # Calcular puntos usando los elementos guardados en la lista 'datos'
          puntos = calcular_puntuacion(datos[2], datos[3], datos[4], datos[5])
          prioridad = clasificar_candidato(puntos)

        # Mostrar resultado individual inmediato
          mostrar_puntaje_individual(datos[0], puntos, prioridad)

        # Guardar en la lista global solo los campos requeridos para el DataFrame final
          candidatos_inscritos.append([datos[0], datos[1], puntos, prioridad])

    # Una vez alcanzados los 10 registros, se bloquea la recepción y se genera el informe
    generar_informe(candidatos_inscritos)
```

```python
#  Función para mostrar el puntaje individual
def mostrar_puntaje_individual(nombre, puntuacion, prioridad):
    print(f"\n--- Resumen Candidato: {nombre} ---")
    print(f"Puntuación Total: {puntuacion}")
    print(f"Clasificación: {prioridad}\n")
```

```python
if __name__ == "__main__":
    main()
```
