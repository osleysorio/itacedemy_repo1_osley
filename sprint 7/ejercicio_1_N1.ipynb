#Ejercicio 1: Selección de Personal

## Algoritmo


## Función obtenir_dades()

- Solicitar al usuario el nombre y apellidos
- Solicitar el teléfono de contacto
- Solicitar los años de experiencia (número entero)
- Solicitar conocimiento de catalán (0-3)
- Solicitar dominio de Excel (0-3)
- Solicitar título académico (0-3)
- Retornar todos los datos como una tupla

```python
def obtenir_dades():
    """Solicita y retorna los datos de un candidato."""
    print("\n--- Nuevo candidato ---")
    nom = input("Nombre y Apellidos: ").strip()
    telefon = input("Teléfono de contacto: ").strip()

    try:
        experiencia = int(input("Años de experiencia en la empresa: "))
        catalan = int(input("Conocimiento de Catalán (0-3): "))
        excel = int(input("Dominio de Excel (0-3): "))
        titulacio = int(input("Título académico (0-3): "))
    except ValueError:
        print("Error: Introduce valores numéricos válidos.")
        return None

    return (nom, telefon, experiencia, catalan, excel, titulacio)
```

## Función validar_dades(dades)

- Verificar que el nombre no esté vacío
- Verificar que el teléfono tenga 9 dígitos
- Verificar que los años de experiencia sean ≥ 0
- Verificar que catalán, excel y titulación estén entre 0 y 3
- Retornar True si todo es válido, False en caso contrario

```python
def validar_dades(dades):
    """Valida que todos los datos sean correctos."""
    if dades is None:
        return False

    nom, telefon, experiencia, catalan, excel, titulacio = dades

    # Validar nombre
    if not nom:
        print("Error: El nombre no puede estar vacío.")
        return False

    # Validar teléfono (9 dígitos)
    if not (len(telefon) == 9 and telefon.isdigit()):
        print("Error: El teléfono debe tener exactamente 9 dígitos.")
        return False

    # Validar experiencia
    if experiencia < 0:
        print("Error: Los años de experiencia no pueden ser negativos.")
        return False

    # Validar puntuaciones (0-3)
    for valor, nombre_campo in [(catalan, "Catalán"), (excel, "Excel"), (titulacio, "Titulación")]:
        if valor < 0 or valor > 3:
            print(f"Error: {nombre_campo} debe estar entre 0 y 3")
            return False
    return True
```

## Función calcular_puntuacio(experiencia, catalan, excel, titulacio)

- Calcular puntuación = experiencia * 1.5 + catalan + excel + titulacio
- Retornar el resultado


```python
def calcular_puntuacio(experiencia, catalan, excel, titulacio):
    """Calcula la puntuación total del candidato."""
    puntuacio = experiencia * 1.5 + catalan + excel + titulacio
    return puntuacio
```

##Funció classificar_candidat(puntuacio)

- Si puntuación ≥ 12 → “Prioridad Alta”
- Si 8 ≤ puntuación < 12 → “Candidato adecuado”
- Si puntuación < 8 → “No prioritario”
- Retornar la clasificación

```python
def classificar_candidat(puntuacio):
    """Clasifica al candidato según su puntuación."""
    if puntuacio >= 12:
        return "Prioridad Alta"
    elif puntuacio >= 8:  # 8 <= puntuacion < 12
        return "Candidato adecuado"
    else:  # puntuacion < 8
        return "No prioritario"
```

##Función mostrar_individual(nom, telefon, puntuacio, classificacio)

- Mostrar por pantalla: nombre, teléfono, puntuación y clasificación

```python
def mostrar_individual(nom, puntuacio, classificacio):
    """Muestra el resultado individual de un candidato."""
    print(f"\n--- Resultado para {nom} ---")
    print(f"Puntuación total: {puntuacio:.1f}")
    print(f"Clasificación: {classificacio}")
    print("-" * 40)
```

##Función mostrar_informe(candidats)

- Ordenar candidatos: primero por prioridad (Alta > Adecuado > No prioritario), - luego por puntuación descendente
- Mostrar título del informe
- Para cada candidato, mostrar nombre, teléfono y puntuación

```python
def mostrar_informe(candidats):
    """Genera y muestra el informe final ordenado."""
    # Definir orden de prioridad
    ordre_prioritat = {"Prioridad Alta": 0, "Candidato adecuado": 1, "No prioritario": 2}

    # Ordenar: primero por prioridad, luego por puntuación descendente
    candidats_ordenats = sorted(candidats, key=lambda c: (ordre_prioritat[c[3]], -c[2]))

    print("\n" + "=" * 60)
    print("INFORME FINAL - CANDIDATOS INSCRITOS")
    print("=" * 60)
    print(f"{'Nombre':<25} {'Teléfono':<12} {'Puntuación':<12} {'Prioridad'}")
    print("-" * 60)

    for nom, telefon, puntuacio, classificacio in candidats_ordenats:
        print(f"{nom:<25} {telefon:<12} {puntuacio:<12.1f} {classificacio}")

    print("=" * 60)
```

##Función main_seleccio()

- Inicializar lista vacía de candidatos
- Mientras haya menos de 10 candidatos:
- a. Llamar a obtenir_dades()
- b. Llamar a validar_dades()
- c. Si datos válidos:
   -Calcular puntuación
   -Clasificar candidato
   -Mostrar resultado individual
   -Añadir a la lista
- d. Si no válidos: mostrar error y repetir
- Cuando haya 10 candidatos, mostrar informe final

```python
def main_seleccio():
    """Función principal que orquesta todo el proceso."""
    candidats = []  # Lista para almacenar (nombre, telefono, puntuacion, clasificacion)

    print("=" * 50)
    print("SISTEMA DE SELECCIÓN DE PERSONAL")
    print("=" * 50)
    print(f"Plazas disponibles: 10")

    while len(candidats) < 10:
        print(f"\nCandidatos inscritos: {len(candidats)}/10")

        # Obtener datos
        dades = obtenir_dades()

        # Validar datos
        if not validar_dades(dades):
            print("Por favor, corrige los errores e intenta de nuevo.")
            continue

        nom, telefon, experiencia, catalan, excel, titulacio = dades

        # Calcular puntuación
        puntuacio = calcular_puntuacio(experiencia, catalan, excel, titulacio)

        # Clasificar candidato
        classificacio = classificar_candidat(puntuacio)

        # Mostrar resultado individual
        mostrar_individual(nom, puntuacio, classificacio)

        # Guardar candidato
        candidats.append((nom, telefon, puntuacio, classificacio))

    # Mostrar informe final cuando hay 10 candidatos
    print("\n" + "!" * 50)
    print("¡LÍMITE ALCANZADO! Se han inscrito 10 candidatos.")
    print("!" * 50)

    mostrar_informe(candidats)

```

```python
# Para ejecutar el programa
if __name__ == "__main__":
    main_seleccio()
```
