## Importar librerias

```python
import random
import string
```

* Solicitar al usuario una entrada numérica de texto para definir la dimensión de la clave.
* Evaluar mediante un bloque de excepciones que la entrada sea un entero. Si no lo es, lanzar una alerta y detener el programa.
* Verificar que la dimensión numérica ingresada sea estrictamente mayor que cero. Si es negativa o nula, abortar la ejecución.
* Desplegar un bucle interactivo que pregunte de forma consecutiva la activación de los juegos de caracteres (Mayúsculas, Minúsculas, Números y Símbolos).
* Restringir la respuesta del usuario mediante lógica condicional, forzando la repetición de la pregunta si el texto no corresponde a "S" o "N".
* Almacenar los estados lógicos verdaderos o falsos de cada opción  y retornarlo.

```python
def obtener_configuracion_usuario():
    #Solicita y valida los datos de entrada del usuario.
    try:
        longitud = int(input("Ingrese la longitud de la contraseña: "))
        if longitud <= 0:
            print("Error: La longitud debe ser un número entero mayor que cero.")
            return None
    except ValueError:

        print("Error: Por favor, ingrese un número entero válido para la longitud.")
        return None

    def pedir_confirmacion(mensaje):
        """Fuerza al usuario a responder únicamente S o N."""
        while True:
            respuesta = input(mensaje).strip().lower()
            if respuesta in ['s', 'n']:
                return respuesta == 's'
            print("Por favor, responda con 'S' para Sí o 'N' para No.")

    mayusculas = pedir_confirmacion("¿Activar mayúsculas? (S/N): ")
    minusculas = pedir_confirmacion("¿Activar minúsculas? (S/N): ")
    numeros = pedir_confirmacion("¿Activar números? (S/N): ")
    caracteres = pedir_confirmacion("¿Activar caracteres especiales? (S/N): ")

    return {
        "longitud": longitud,
        "may_activo": mayusculas,
        "min_activo": minusculas,
        "num_activo": numeros,
        "car_activo": caracteres
    }
```

* Mapear los grupos de caracteres estándar de la librería string con sus banderas booleanas de activación correspondientes.
* Filtrar e integrar en una lista únicamente aquellos repertorios que el usuario haya seleccionado con un estado "Sí".
* Comprobar que la lista de conjuntos activos no se encuentre vacía. En caso de estarlo, retornar un mensaje de error explícito.
* Validar que la longitud deseada sea igual o mayor al número de conjuntos activos, asegurando espacio matemático para cumplir los requisitos de variedad.* Extraer al azar un carácter obligatorio de cada tipo seleccionado para garantizar el cumplimiento de las políticas de seguridad.
* Unificar los alfabetos de todos los conjuntos activos en una única cadena masiva de caracteres.
* Calcular la longitud restante del vector y rellenar las posiciones faltantes tomando muestras aleatorias con repetición de la cadena unificada.
* Concatenar los caracteres de presencia obligatoria con la sublista de relleno aleatorio.
* Reordenar de manera aleatoria el conjunto final mediante un muestreo posicional completo sin repetición para disolver el orden de inserción original.
* Convertir la lista de caracteres resultante en una sola cadena de texto compacta y retornarla.

```python
def crear_pass(longitud, may_activo=True, min_activo=True, num_activo=True, car_activo=True):
    """Genera la contraseña basándose en los parámetros de configuración."""
    opciones = [
        (string.ascii_uppercase, may_activo),
        (string.ascii_lowercase, min_activo),
        (string.digits, num_activo),
        (string.punctuation, car_activo)
    ]


    sets_activos = [chars for chars, activo in opciones if activo]

    if not sets_activos:
        return "Error: Debes activar al menos un tipo de carácter."

    if longitud < len(sets_activos):
        return f"Error: La longitud debe ser de al menos {len(sets_activos)} caracteres para los tipos seleccionados."

    # PASO 1: Asegurar estrictamente un carácter de cada tipo seleccionado
    #password_obligatorio = [random.choice(chars) for chars in sets_activos]
    password_obligatorio = [random.choice(chars) for chars in sets_activos]

    # PASO 2: Rellenar el resto de la longitud de forma aleatoria
    todos_los_caracteres = "".join(sets_activos)
    longitud_restante = longitud - len(password_obligatorio)
    password_restante = [random.choice(todos_los_caracteres) for _ in range(longitud_restante)]

    # PASO 3: Combinar ambos bloques
    password_total = password_obligatorio + password_restante

    # PASO 4: Mezclar usando 'random.sample' (más seguro que shuffle en funciones modulares)
    # Al pedirle la misma longitud del total, reordena todos los elementos sin perder ninguno
    password_mezclado = random.sample(password_total, len(password_total))

    return "".join(password_mezclado)
```

* llammar la rutina de configuración de parámetros del usuario.
* Evaluar si el retorno de la configuración contiene datos válidos; si devuelve un valor nulo, finalizar el flujo del algoritmo de inmediato de forma controlada.
* Transferir las variables de configuración almacenadas como argumentos funcionales hacia el motor de construcción de la contraseña.
* Imprimir  la cotraseña generada.



```python
def ejecutar_generador():
    """Controla el flujo principal del programa."""
    config = obtener_configuracion_usuario()

    # Si la configuración falló por un error de tipo (ValueError), frena la ejecución
    if config is None:
        return

    nueva_contrasena = crear_pass(
        longitud=config["longitud"],
        may_activo=config["may_activo"],
        min_activo=config["min_activo"],
        num_activo=config["num_activo"],
        car_activo=config["car_activo"]
    )

    print(f"Contraseña generada: {nueva_contrasena}")
```

```python
# Punto de entrada oficial del script
if __name__ == "__main__":
    ejecutar_generador()
```
